```python
# [1번째 셀] 부동산 매물 데이터 및 나의 조건 세팅

매물_리스트 = [
    {
        "이름": "A매물 (봉천역 원룸)",
        "월세": 45,
        "방음": "불량",
        "치안": "보통",
        "수압": "약함"
    },
    {
        "이름": "B매물 (강남역 오피스텔)",
        "월세": 75,
        "방음": "우수",
        "치안": "안전",
        "수압": "좋음"
    },
    {
        "이름": "C매물 (대학가 신축 빌라)",
        "월세": 48,
        "방음": "우수",
        "치안": "안전",
        "수압": "좋음"
    }
]

내_최대_예산 = 50  # 50만 원 이하만 가능
print("✅ [1단계 완료] 방 정보와 나의 조건(월세 50만 원 이하, 방음 필수, 치안 안전) 입력 완료!")
```

```python
# [2번째 셀] 조건에 맞는 정답 매물 자동 찾기 (AI 가상 분석)

AI_추천_결과 = None

# 매물 리스트를 하나씩 돌면서 조건에 맞는지 자동으로 검사합니다.
for 방 in 매물_리스트:
    if 방["월세"] <= 내_최대_예산 and 방["방음"] == "우수" and 방["치안"] == "안전":
        AI_추천_결과 = 방
        break

print("🤖 AI 분석 완료!")
print(f"👉 내 조건에 딱 맞는 정답 매물은 바로 **[{AI_추천_결과['이름']}]** 입니다.")
```

```python
# [3번째 셀] 분석 결과 자동 구조화 및 표 출력
import pandas as pd

AI_정리_보고서 = {
    "최종 추천 매물": AI_추천_결과["이름"],
    "월세 가격": f"{AI_추천_결과['월세']}만 원 (예산 통과)",
    "방음 및 치안": f"방음 {AI_추천_결과['방음']} / 치안 {AI_추천_결과['치안']} (조건 만족)",
    "A매물 탈락 이유": "월세는 맞지만 수압이 약하고 방음이 안 됨",
    "B매물 탈락 이유": "방은 좋으나 월세(75만 원)가 내 예산을 초과함"
}

# 표 형태로 변환하여 화면에 자동 출력
표 = pd.DataFrame(list(AI_정리_보고서.items()), columns=["분석 항목", "AI 자동 분석 내용"])
표
```

```python
# [4번째 셀] 시스템 최종 자동 검증

오류_체크 = 0

print("🔍 추천된 매물에 오류가 없는지 최종 안전 검증을 시작합니다...")
print("-" * 60)

# 최종 추천된 방의 월세 검증
if AI_추천_결과["월세"] > 내_최대_예산:
    print(f"❌ [오류 발견] 추천된 방의 월세({AI_추천_결과['월세']}만 원)가 예산을 초과했습니다!")
    오류_체크 += 1
else:
    print(f"✅ [검증 통과] 월세 가격 {AI_추천_결과['월세']}만 원으로 내 예산(50만 원) 한도 이내입니다.")

# 최종 추천된 방의 방음 검증
if AI_추천_결과["방음"] != "우수":
    print("❌ [오류 발견] 조용한 방음 조건에 맞지 않는 방입니다!")
    오류_체크 += 1
else:
    print("✅ [검증 통과] 재택근무에 적합한 조용한 방음 상태 확인 완료.")

# 최종 결론 출력
print("-" * 60)
if 오류_체크 == 0:
    print(f"🎉 [최종 통과] 자동 시스템 검증 성공! {AI_추천_결과['이름']}은 완벽한 정답입니다. 계약하러 가셔도 좋습니다!")
else:
    print("⚠️ [검증 실패] AI 분석 결과에 조건 오류가 발견되었습니다.")
```
